trying to read the raw data works fine in `pandas`, but causes errors in `polars`.
For some reason, when a value is `NaN` in polars it breaks when reading the data.
So, we preprocess the data to convert nan values to `None` and then read it in.

```python
from pathlib import Path
import math

import numpy as np
import polars as pd

# muutils
from muutils.jsonlines import jsonl_write, jsonl_load
from muutils.dbg import dbg, dbg_tensor
from muutils.tensor_info import array_summary
```

```python
scalar_feats_raw = jsonl_load(Path("../../data/scalar_features.jsonl"))
```

```python
len(scalar_feats_raw)
```

```python
x = scalar_feats_raw[0]["feat.markov_transition.time.skewness"]
print(x)
print(type(x))
math.isnan(x)
```

```python
scalar_feats_nan_to_null = [
	{
		k: (
			v 
			if not (isinstance(v, float) and math.isnan(v)) 
			else None
		)
		for k, v in s.items()
	}
	for s in scalar_feats_raw
]
```

```python
jsonl_write(Path("../../data/scalar_features_fixed.jsonl"), scalar_feats_nan_to_null)
```
